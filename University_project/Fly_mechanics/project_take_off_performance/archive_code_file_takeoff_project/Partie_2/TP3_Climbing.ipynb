# **<span style="color:red">A - Calcul de performance en montée</span>**

<hr style="border:2px solid orange " width="99%">

## **<span style="color:blue"> I - Angle de montée</span>**
Considérons un avion en phase de montée (fig. ci-dessous).

 

| <img src="Images/climbing.png" width="600">  |
|:--:|
|**Fig.1 Avion en montée [1]**|
 

En négligeant l'angle $\varepsilon$ on suppose que la force de propulsion $T$ est alignée avec le vecteur vitesse de l'avion $V$ et on vérifie les équations habituelles sous l'hypothèse d'une vitesse $V$ **constante**:

$$
L - W~\cos \theta = 0 \qquad \qquad T = D - W~\sin \theta
$$
qu'on met sous la forme:

$$
L = W~\cos \theta, \qquad \qquad \sin \theta = \dfrac{T - D}{W}
$$
$\theta$ est **l'angle de montée**.


**Approximation**:

Dans la suite comme les angles $\theta$ sont faibles on approxime toujours $\cos\theta \approx 1$


<div class="alert alert-block alert-danger">
<b> Angle de montée $\theta$</b>
$$
\theta = \arcsin \dfrac{T - D}{ W}
$$
</div>


## **<span style="color:blue"> II - Vitesse ascentionnelle, *Rate of Climb* </span>**

En mécanique on travaille souvent en puissance et on applique le théorème de l'énergie cinétique:

 
<div class="alert alert-block alert-info">
<b> Théorème de l'énerge cinétique </b>
$$
\dfrac{d}{dt} \left(\dfrac{W}{2~g}~V^2 \right) + \dfrac{d}{dt} \left( W~h \right) = P_a - P_r
$$
</div>



Les notations de l'équation précédente sont indiquées dans la table 1.


|                      Sens      |  Equation, modèle   | Unités    | Anglais          |  
|:------------------------------:|:-------------------:|:---------:|:-----------------|           
| poids                          | $W = m~g$           | N         | weight           |
| altitude                       | $h $                | m, km, ft | altitude         | 
| puissance disponible           | $P_a = T \times V$  | W, kW, hp | available power  |
| puissance requise (nécessaire) | $P_r = D \times V$  | W, kW, hp | required power   |
 
|    | 
|:--:|
|**Table 1  Notations de l'équation du théorème de l'énergie cinétique**|


Après dérivation on aboutit à :
$$
 \dfrac{V}{g}~\dfrac{dV}{dt} + V_c =  \dfrac{P_a - P_r}{W} \tag{eq. 1}
$$

* La vitesse ascentionnelle $V_c$ ou *Rate of Climb* est alors:

<div class="alert alert-block alert-danger">
<b> Définition de la vitesse ascentionnelle </b>
 $$V_c = ROC = \dfrac{dh}{dt} = V~\sin \theta$$
</div>

  Dans la littérature pour ROC on trouve aussi la notation R/C.


* l'excès de puissance $P_e$ est donné par:


<div class="alert alert-block alert-success">
<b> Calcul  l'excès de  puissance </b>
$$
P_e = P_a - P_r = V~(T - D)
$$
</div>


L'équation (1) est fondamentale: elle indique qu'en fonction de l'excès de puissance l'avion peut accélérer ou monter ou faire les deux à la fois.

Dans la suite on suppose une montée à vitesse constante et on écrit:

<div class="alert alert-block alert-success">
<b> Calcul de la vitesse ascentionnelle </b>
 $$
V_c = ROC =  \dfrac{P_a - P_r}{W} \hspace{4cm} \sin \theta = \dfrac{V_c}{V} =  \dfrac{T - D}{W} \tag{eq. 2}
$$
</div>



| <img src="Images/airspeed.png" width="400">  |
|:--:|
|**Fig.2 Composantes de la vitesse, vitesse ascentionnelle [1]**|

 

 

L'équation (2) permet le calcul de la vitesse ascentionnelle et de l'angle de montée, en fonction du type de propulsion et des caractéristiques structurelles et aérodynamiques de l'avion.


### Unité de la vitesse ascentionnelle

* La ROC est donnée en aéronautique en ft/mn, dont l'abbréviation est **fpm**
* Dans le système métrique on la donne en m/mn ou m/s

# **<span style="color:red">B - Modèles en propulsion </span>**

<hr style="border:2px solid orange " width="99%">


* Les modèles physiques et mathématiques sont fonction du type de propulsion et en particulier de la technologie des moteurs.

* On présente ici les modèles indiqués dans [1], sans nécessairement entrer dans tous les détails.

* Dans la suite la force de propulsion ou la puissance de référence sont prises au niveau de la mer
et indicée $sl$ pour *sea level*.


## **<span style="color:brown">I - Moteurs à pistons </span>**

## **<span style="color:blue">a - Modèle simple </span>**

C'est le premier modèle, pour des moteurs thermiques basiques (à pistons/cylindres).

<div class="alert alert-block alert-info">
<b> Engine model : simple </b>
$$
P_a = P_{sl}~\sigma, \qquad \qquad \sigma = \rho(h)~/~\rho_{sl} \tag{Simple model}
$$
</div>



## **<span style="color:blue">b - Modèle  de Gagg et Ferrar </span>**

Ce modèle prend mieux en compte l'effet de changement de température en fonction de l'altitude

<div class="alert alert-block alert-info">
<b> Engine model : "Gagg and Ferrar" </b>
$$
P_a = P_{sl}~\left( \sigma - \dfrac{1-\sigma}{7.55} \right) =  P_{sl}~\left( 1.132 \sigma -0.132  \right) ,  \tag{GF model}
$$
</div>




## **<span style="color:blue">c - Modèle d'un moteur thermique avec turbocompression (*turbocharger*) </span>**

| <img src="Images/turbocharger.png" width="500">  |
|:--:|
|**Fig.3 Moteur turbocompressé [2]**|
 

On peut augmenter la puissance d'un moteur avec un turbocompresseur qui permet de récupérer de la puissance supplémentaire à partir des gaz de combustion. Le modèle associé est basé sur celui de Gagg et Ferrar en introduisant une altitude critique $h_c$:
 
<div class="alert alert-block alert-info">
<b> Engine model :  turbocharger </b>
 
* si $h \leqslant h_c$:
$$
P = P_{sl}
$$
    
* sinon:
$$
P = P_{sl} \left(1.132~( 1 - \kappa~(h-h_c))^{4.2561} - 0.132 \right)  \tag{Turbocharger model}
$$
</div>

Le terme en $\kappa$ est le modèle simplifié de l'atmosphère standard qui permet de prendre en compte la variation de masse volumique en fonction de l'altitude. Il a déjà été utilisé dans le TP n°1.


## **<span style="color:blue">d - Prise en compte de la température réelle en altitude </span>**


Le modèle d'atmosphère standard fournit une température de référence notée aussi $T_{ISA}$ ou $T_{std}$ (std pour standard). 

Le fonctionnement réel du moteur doit prendre en compte la température réelle en altitude en fonction de la météo.

En anglais on l'appelle *Outside Atmosphere Temperature* ou plus simplement l'**$OAT$** ou ici $T_{OAT}$.

On note la variation de la température $OAT$ à celle de l'atmosphère de référence ISA :

$$
\Delta T_{OAT} = T_{OAT} - T_{std}
$$

* La vraie puissance est alors supposée de la forme :

<div class="alert alert-block alert-info">
<b> Engine model :  piston + correction 1 </b>
$$
\dfrac{P_a}{P_{std}} = \sqrt {\dfrac{T_{std}}{T_{OAT}}} = \sqrt{ \dfrac{273.15~( 1 - \kappa~h)}{T_{OAT}}}    \tag{Correction 1}
$$
</div>


Cette correction est approximative puisqu'on suppose une dépendance en racine carré avec la température.

* On peut appliquer une autre correction, peut-être plus judicieuse. Elle consiste à recalculer la masse volumique à la bonne altitude avec la loi d'état $\rho = p  / (r T)$ puis d'appliquer le modèle $GF$:



<div class="alert alert-block alert-info">
<b> Engine model :  piston + correction 2 </b>
 $$
\sigma = \dfrac{\rho}{\rho_{sl}} = \dfrac{p}{p_{sl}} \times \dfrac{T_{sl}}{T_{OAT}} =  \delta(h)  \times \dfrac{T_{sl}}{T_{OAT}}  \tag{Correction 2}
$$
</div>


Le rapport des pressions statiques $\delta(h)$ est donné par le modèle atmosphérique (ou mesuré dans la réalité).

## **<span style="color:blue">e - Autres modèles </span>**

* Comme pour tous les moteurs à pistons, la puissance dépend de la vitesse de rotation, du nombre de pistons et de la cylindrée,
donc on peut améliorer les modèles précédents

* On donne ici la puissance à la sortie du moteur, on pourrait prendre en compte le rendement des réducteurs, l'effet du refroidissement.

* Enfin la puissance pour la propulsion est définie à partir de l'hélice.

On a donc matière à améliorer le modèle "moteur" (se reporter à [1]).

par la turbomachine sert à faire tourner une hélice propulsive, et une légère poussée résiduelle améliore le rendement propulsif.



|<img src="Images/turboprop_modeles.png" width="500">|

## **<span style="color:blue">a - Modèle Mattingly</span>**
  

* Le modèle est mixte, il prend en compte la puissance associée à l'hélice mais aussi le "jet" supplémentaire qui provient
du moteur à réaction qui fait tourner l'arbre moteur.

* Comme pour les avions à réactions c'est la force de propulsion $T$ qui est modélisée au lieu de la puissance pour les moteurs à pistons.



* Cependant dans cette section on utilise la lettre $T$ pour la température, la force de propulsion est $F$.

* Nous présentons, de [1], le modèle le plus simple.

Le fonctionnement du moteur est fortement lié à la pression et température ambiante, quelque soit l'altitude, et à la pression totale ou pression isentropiqe $p_i$ (notée aussi $p_t$) et à la température totale  (ou isentropique) $T_i$ (notée aussi $T_t$).

On pourra se reporter au TP n°1 pour se rappeler ces différents notions.

Ainsi on a :

$$
\dfrac{T_t}{T } = 1 + \dfrac{\gamma -1}{2}~M^2 
$$


|<img src="Images/pi_c.png" width="900">|
|:--:|
|**Fig.7 Taux de compression et $\theta_0$ pour un turbopropulseur, température en degrés Rankine [3]**|


Cette relation intervient dans le calcul du taux de compression $\pi_c$ entre l'entrée d'air et la sortie de la chambre de combustion, qui influence fortement la force de poussée.

Sa valeur maximale, fonction de l'altitude, de la vitesse et de la température totale à la sortie de la chambre de combustion $T_{t_4}$  est appelé **theta-break** (cf fig. 7). $\theta_0$ est défini juste dessous.

La valeur de fonctionnement de ce coefficient pour un moteur est appelé le "**Throttle Ratio**" noté $T\!R$, et est optimisé pour la vitesse de croisière à une altitude donnée. Les détails de l'étude se trouvent dans l'annexe D de [1]. Les calculs sont assez complexes.

Le modèle de propulsion fait alors appel à quatre étapes de calcul, fournie par Mattingly et al dans [2]

1. Calcul de la force de propulsion au niveau de la mer $F_{sl}$ (ou récupération des données constructeur)
2. Calcul du rapport des températures en fonction de l'altitude ou de la température réelle $T = T_{OAT}$:

<div class="alert alert-block alert-info">
<b> Engine model :  $\theta_0$ </b>
$$
\theta_0 = \dfrac{T_t}{T_0} =  \dfrac{T}{T_0}~\left(1 + \dfrac{\gamma -1}{2}~M^2\right)  \tag{Theta-0} 
$$
</div>

$T_0$ est la température de l'atmosphère au niveau du sol. Comme vu dans le TP n°1 on a noté
$$\Theta(h) = \dfrac{T_{std}(h)}{T_0}$$

$T_t$ est la température totale donnée dans les conditions de vol (à l'altitude $h$).

3. Calcul du rapport des pressions:


<div class="alert alert-block alert-info">
<b> Engine model :  $\delta_0$ </b>
$$
\delta_0 = \dfrac{p_t}{p_0} = \dfrac{p}{p_0}~\left(1 + \dfrac{\gamma -1}{2}~M^2\right)^{\frac{\gamma}{\gamma-1}}
$$
</div>

$p_0$ est la pression de l'atmosphère au niveau de la mer ($p_{sl})$.
Comme vu dans le TP n°1 on a noté
$$\delta(h) = \dfrac{p(h)}{p_0}$$

$p$ sinon c'est la pression statique mesurée dans les conditions de vol à l'altitude $h$.

$p_t$ est la pression totale dans les conditions de vol.

4. Dernière étape, calcul de la poussée $F$:
    <div class="alert alert-block alert-info">
    <b> Engine model :  turboprop </b>
    
    * si $M\leqslant 0.1$ :
    $$ F = F_{sl}~\delta_0$$
    * sinon :
        * si $\theta_0 \leqslant T\!R$:
        $$ F = F_{sl}~\delta_0~\left[1 - 0.96~(M-0.1)^{\frac{1}{4}}  \right]$$
        * sinon
        $$ F = F_{sl}~\delta_0~\left[1 - 0.96~(M-0.1)^{\frac{1}{4}} - \dfrac{3(\theta_0 - T\!R)}{M-0.1}  \right] \tag{turboprop model}$$
    </div> 
    * La valeur courante pour $T\!R$ est 1.072
            
## **<span style="color:blue">b - Autres modèles</span>**            
 

On peut trouver d'autres modèles, fournis parfois par les constructeurs. Un autre modèle courant porte sur la puissance disponible

$$
P = P_{sl} \times A \times M^n,
$$
L'exposant est autour de 1/2 et la constante est fournie par le constructeur.

Ce dernier modèle n'est pas implanté dans le code conçu pour ce TP.

## **<span style="color:brown">III - Moteurs à réactions simple flux (*turbojet*)</span>**

<hr style="border:2px solid orange " width="99%">

|<img src="Images/turbojet.png" width="500">|
|:--:|
|**Fig.5 Turboréacteur simple flux [2]**|

* Ce sont les réacteurs d'autrefois peu utilisé de nos jours, à cause de leur consommation élevée et du niveau élevé de bruit émis.

* Le modèle suit le modèle du turbopropulseur ce qui est logique, puisque le coeur des moteurs ont le même fonctionnement. 

* Mattingly [3] fournit la démarche en quatre temps. Pour les étapes 3 et 4 on a deux possibilités, avec ou sans le fonctionnement militaire, à savoir l'emploi de la post-combustion.

Voici les étapes:
1. Calcul de la force de propulsion au niveau de la mer $F_{sl}$ (ou récupération des données constructeur)
2. Calcul du rapport des températures en fonction de l'altitude ou de la température réelle $T = T_{OAT}$:
$$
\theta_0 = \dfrac{T_t}{T_0} =  \dfrac{T}{T_0}~\left(1 + \dfrac{\gamma -1}{2}~M^2\right)
$$
2. Calcul du rapport des pressions:
$$
\delta_0 = \dfrac{p_t}{p_0} = \dfrac{p}{p_0}~\left(1 + \dfrac{\gamma -1}{2}~M^2\right)^{\frac{\gamma}{\gamma-1}}
$$
4. Calcul de la poussée maximale
    * Avec post-combustion :
        <div class="alert alert-block alert-info">
        <b> Engine model :  turbojet with afterburners </b>
        - si $\theta_0 \leqslant T\!R$:
        $$ F = F_{sl}~\delta_0~\left[1-0.3~(\theta_0-1)  - 0.1~\sqrt{M}\right]$$
        - sinon
        $$ F = F_{sl}~\delta_0~\left[1-0.3~(\theta_0-1)  - 0.1~\sqrt{M}- \dfrac{1.5~(\theta_0-T\!R)}{\theta_0}\right]$$
          </div> 
          
    * Sans post-combustion (poussée dite militaire)
    <div class="alert alert-block alert-info">
    <b> Engine model :  turbojet without afterburners (military thrust) </b>

        - si $\theta_0 \leqslant T\!R$:
        $$ F = 0.8~F_{sl}~\delta_0~\left[1 - 0.16~\sqrt{M}\right]$$
        - sinon
    $$ F = 0.8~F_{sl}~\delta_0~\left[1 - 0.16~\sqrt{M} - \dfrac{24~(\theta_0-T\!R)}{(M + 9)~\theta_0}\right]$$
     </div> 
     


## **<span style="color:brown">IV - Moteurs à réactions double  flux (*turbofan*)</span>**

|<img src="Images/turbofan2.png" width="700">|
|:--:|
|**Fig.6 Turboréacteur double flux [2]**|

Dans le cas du réacteur double flux, le flux primaire est le flux principal qui suit le cycle classique d'un turboréacteur simple. Cependant à l'entrée, un flux secondaire traverse une roue (un fan) qui ne passe pas dans le compresseur, puis la chambre de combustion. On définit alors **le taux de dilution** ou **bypass ratio** noté BPR qui est le rapport 
entre le flux secondaire et le flux primaire. Plus il est grand et plus le flux secondaire est important.

On distingue les réacteurs à faible taux de dilution (low BPR), en gros inférieur à 3.5 et les réacteurs à fort taux de dilution, au delà de 4, voir 7 et plus (high BPR).  

Le modèle doit prendre en compte cela. On trouve la formulation dans [2] et [1]:

1. Calcul de la force de propulsion au niveau de la mer $F_{sl}$ (ou récupération des données constructeur)
2. Calcul du rapport des températures en fonction de l'altitude ou de la température réelle $T = T_{OAT}$:
$$
\theta_0 = \dfrac{T_t}{T_0} =  \dfrac{T}{T_0} \left(1 + \dfrac{\gamma -1}{2}M^2\right)
$$
2. Calcul du rapport des pressions:
$$
\delta_0 = \dfrac{p_t}{p_0} = \dfrac{p}{p_0} \left(1 + \dfrac{\gamma -1}{2}M^2\right)^{\frac{\gamma}{\gamma-1}}
$$
4. Calcul de la poussée maximale
    * Réacteur à faible taux de dilution (low BPR) 
         <div class="alert alert-block alert-info">
         <b> Engine model :  turbofan, low BPR </b>
    
        * Avec post-combustion:
            - si $\theta_0 \leqslant T\!R$:
            $$ F = F_{sl} \delta_0 $$
            - sinon
            $$ F = F_{sl} \delta_0 \left[1 - \dfrac{3.5 (\theta_0-T\!R)}{\theta_0}\right]$$

        * Sans post-combustion (poussée dite militaire): 
            - si $\theta_0 \leqslant T\!R$:
            $$ F = 0.6 F_{sl} \delta_0 $$
            - sinon
             $$ F = 0.6 F_{sl}  \delta_0 \left[1 -\dfrac{3.8 (\theta_0-T\!R)}{\theta_0}\right]$$
          </div> 
          
       * Réacteurs à fort taux de dilution (high BPR) 
         <div class="alert alert-block alert-info">
         <b> Engine model :  turbofan, high BPR </b>
    
            - si $\theta_0 \leqslant T\!R$:
             $$ F =  F_{sl} \delta_0 \left[1 - 0.49 \sqrt{M}\right]$$
            - sinon
             $$ F =  F_{sl} \delta_0 \left[1 - 0.49 \sqrt{M} - \dfrac{3 (\theta_0-T\!R)}{M + 1.5}\right]$$
         </div> 
         
         
## **<span style="color:brown">V - Conclusions</span>**

Les différents modèles proposés ici sont des versions simplifiées de modèles beaucoup plus complexes qui prennent en compte 
la technologie des propulsions, et en particulier les différents cycles thermodynamiques, le nombre et type de compresseurs ou de turbine, les températures et pressions totales atteintes dans chacun des cycles, etc ...

Les références [2] et [3] précisent beaucoup mieux la complexité de la modélisation. 

Les différents modèles présentés ici sont implémentés dans le module Python fourni pour le TP.
