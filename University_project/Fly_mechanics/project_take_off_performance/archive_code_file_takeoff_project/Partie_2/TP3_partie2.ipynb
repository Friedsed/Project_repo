## **<span style="color:magenta">a- Comment utiliser les fonctions et entrer des cas ? </span>**

### Principe de la programmation

Le module ``flight.testcases`` contient les fonctions qui permettent de définir les cas d'étude.
Chaque avion se caractérise par 4 sous-systèmes:

1. des données structures (masse, surface de l'aile, etc ...) obtenues par la fonction  ```get_aircraft(case=k)```
2. des données de vol (domaine de Mach et plage d'altitude)  obtenues par la fonction  ```get_flight_data(case=k)```
3. des données aérodynamiques, essentiellement la polaire,  obtenues par la fonction  ```get_polar_data(case=k)```
4. des données moteurs (type de moteur et options en fonction du modèle) obtenues par la fonction  ```get_polar_engine(case=k)```
    
* Ainsi un avion est constitué d'une liste de 4 entiers correspondant au 4 choix, dans cet ordre

        1. de la structure
        2. du vol
        3. de l'aérodynamique
        4. des moteurs.

* par exemple [0, 0, 0, 0] est l'avion avec tous les premiers cas des 4 sous-systèmes.

* Il est alors facile de prendre le même avion en changeant juste le moteur, par exemple [0, 0, 0, 2]

* Enfin il est facile de rajouter des cas dans les fonctions du module ``flight.testcases``, sur la  base de ce qui existe déjà.

* Chaque fonction renvoie un dictionnaire qu'il faut fournir à une nouvelle instance de la classe ```airplane``` qui définit l'avion. Le mieux pour comprendre est de tester et analyser l'exemple ci-dessous. Des commentaires dans le code expliquent les différentes actions. 
* Ce morceau de code pourra alors être recopié pour un nouvel avion. 
* Informatiquement on peut faire plus cours, mais on a laissé cette manière très descriptive par nécessité pédagogique.

### Premières partie du code : introduire les paramètres d'entrées

````python
# Import de base nécessaire:
import matplotlib.pyplot as plt
from flight.airplane import Airplane
from flight.aircraft import Flight, Aircraft
from flight.quadratic_polar import QuadraticPolar
from flight.testcases import get_aircraft, get_flight_data, get_polar_data, get_engine
````

````python
# définition de plusieurs avions différents. 
cases = ([0, 0, 0, 0], [0, 1, 0, 1], [0, 2, 0, 2], [0, 3, 0, 3], [0, 4, 0, 4], [1, 4, 1, 5])

cas = cases[0]  # je choisis le premier avion de la liste.
# ou bien écrire cas = [0, 0, 0, 0] par exemple
# On suit les 4 étapes : structure, vol, aérodynamique, motorisation
# step 1: enter structure
avion_str = Aircraft(get_aircraft(cas[0]))
# step 2: flight_data
flight = Flight(get_flight_data(cas[1]))
print(" k  :", avion_str.k)     # calcul automatique à partir de l'allongement (donnée structure)
# step 3: aerodynamics/polar
polar_values = get_polar_data(cas[2])
# la polaire est définie dans une classe à part, déjà utilisée dans un TP précédent.
polar = QuadraticPolar(d=dict(k=avion_str.k, cd_0=polar_values[0], cl_0=polar_values[1]))
print(avion_str) # tableau récapitulatif
# fin des choix. On passe à la construction du système
````

**Commentaires**

* Dans la structure il y a une option pour entrer les vitesses, qui n'est pas utilisée ici, donc les vitesses sont à zéro.
Elle sont fixées dans la partie "vol", non affichée ici.

* 0n remarque la rapport $W/S$ qui est assez important, autour de 320 kg/m$^2$ d'aile sur cet exemple.

### Seconde partie: créer l'avion et calculer ses performances en montée

````python
# class instantiation of any aircraft : on démarre la construction de notre modèle avion.
a = Airplane()  # l'avion 'système' est vide pour le moment
a.set_structure_parameters(avion_str)                           # <== step 1
a.set_flight_parameters(flight)                                 # <== step 2
a.set_polar(polar)                                              # <== step 3
a.set_required_power()                                          # calcul de Pr
# step 4: engine choice
a.set_engine(get_engine(cas[3]))                                # <== step 4
a.set_available_power()                                         # calcul de Pa
a.set_climbing_performance()                                    # calcul de Vc
````

**Remarques**

* On retrouve les options du moteur, vérification utile et importante et il faut se souvenir que certaines options ne s'appliquent qu'à certains modèles.
* On parlera plus loin des paramètres à entrer pour le moteur (déjà indiqué par ailleurs dans la partie 1).
* On a un tableau qui nous donne, en fonction de l'altitude la vitesse minimale et maximale de l'avion ainsi que la vitesse ascentionnelle maximale. 
* On est bon sur nos paramètres quand on a peu de lignes où les vitesses sont mise à 0, qui signifie, que l'avion n'a pas assez de puissance ou de force de propulsion pour monter ou rester à son niveau de vol et $Vc$ est alors négative.
* Différentes méthodes ont été implémentées pour calculer les plafonds de service et maximum, notamment par interpolation linéaire ou quadratique. 
* On est bon sur ces dernière données si les valeurs sont cohérentes entre elle et avec l'avion considéré.

### Dernière étape: graphiques des sorties importantes

**Passons maintenant à l'affiche des résultats**

* En plus des affichages produits, on peut avoir des représentation sous forme de courbes.
* Le choix se fait par un dictionnaire, et on met l'option à  "True" pour l'affichage.
* Par défaut, l'abscisse des diagrammes est la vitesse en km/h. 
* Pour avoir le nombre de Mach, il faut mettre "mach=True". C'est l'option à privilégier pour les avions avec turboréacteurs.

Regarder et tester les exemples suivants:

````python
a.plot(choice=dict(perfo=True, force=False, power=False, vc=False), mach=False)
````

* On a ici:
1. l'angle de montée $\sin\theta$
2. la vitesse ascentionnelle $V_c$ ou (ROC),
3. l'angle de montée en degrés
4. L'excès de puissance.

L'aire sous la courbe de l'excès de puissance permet de visualiser la quantité disponible pour accélérer ou monter plus vite.

Les vitesses minimales et maximales sont données pour $V_c=0$ et dépendent de l'altitude.

````python
a.plot(choice=dict(perfo=False, force=False, power=False, vc=True), mach=False)
````

1. Graphe supérieur:

    * Le premier graphe représente la vitesse ascensionnelle en fonction de l'altitude. Dans la théorie c'est une droite.
    * Ici on a une régression linéaire et quadratique. Le comportement varie avec la motorisation.
    * Le coefficient $r$ de la régression linéaire est donnée dans les affichages du code (voir plus haut). Pour une droite on a $|r|=1$
    * Lorsque la courbe atteint zéro, on obtient le plafond maximal
    * le plafond de service correspond à une vitesse ascentionnelle de 100 fpm (à traduire en m/mn).
    * les altitudes sont écrites en pieds (ft) l'unité habituelle du monde de l'aviation;
      ou en mètres (plus compréhensible pour nous).
      
 
2. Graphe inférieur:
    
    * des courbes tracées plus haut on a extrait les vitesses minimales et maximales.
    * lorsque les deux courbes se rejoignent on a atteint le plafond maximal de l'avion

````python
a.plot(choice=dict(perfo=False, force=True, power=False, vc=False), mach=True)
````

**Remarques**

* Les deux forces de propulsion en pointillés, et de traînée (trait continu) sont représentées.
* Pour avoir une vitesse ascentionnelle positive il faut que $T > D$. 
* On trouve bien une vitesse ascentionnelle positive dans des gammes de Mach entre 0.1 et 0.3
* Le graphe est plus lisible si on ne trace qu'une ou deux courbes.
* Pour les très petites vitesses, la polaire réelle redescend vers 0, et donc aussi la force de traînée. Ce n'est pas modélisé ici.

````python
a.plot(choice=dict(perfo=False, force=False, power=True, vc=False), mach=True)
````

**Remarques**

* Le diagramme représente maintenant les puissances. 
* en pointillé la puissance disponible par la motorisation et en trait continu la puissance requise.
* Pour avoir une vitesse ascentionnelle positive il faut que $P_a > P_r$. 
* On trouve bien une vitesse ascentionnelle positive dans des gammes de Mach entre 0.1 et 0.3
* Le graphe est plus lisible si on ne trace peu de courbes
* Cette figure est ici plus lisible que la figure précédente, en effet pour l'avion à moteur à piston, la puissance disponible est quasi-indépendante de la vitesse pour une altitude donnée
